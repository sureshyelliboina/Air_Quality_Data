```python
# Chart 3: Scatter Plot (Distribution of PM2.5 levels over Date)
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv('Air_Quality.csv')
plt.figure(figsize=(8, 4))
plt.scatter(df['Date'], df['PM2.5_Level'], color='purple', s=140)
plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#char 5
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].mean().plot(kind='bar', color='blue')
plt.title('char 5 :Average PM2.5 Levels by Location')
plt.xlabel('Peak PM2.5_Level')
plt.ylabel('Location')
plt.show()
```

```python
  # Chart 5: Vertical Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].max().plot(kind='bar', color='blue')
plt.title('Chart 5: Max PM2.5 Recorded per Location')
plt.xlabel('Location')
plt.ylabel('Max PM2.5')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].mean().plot(kind='barh', color='blue')
plt.title('Chart: Average PM2.5 Levels by Location')
plt.xlabel('Average PM2.5 Level')
plt.ylabel('Location')
plt.show()
```
